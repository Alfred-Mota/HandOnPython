### Perceptron

A unidade basica de uma rede neural. A seguir é apresentado um TLU (threshold logic unit), que aplica uma função degrau dada um conjunto de entradas cada uma com seus pesos.

<img src="./image2.png" width="500">

O neurorio realiza um produto vetorial para ativar o limiar:

$${z = W^T * X + b}$$

Em que:
* W^T é o vetor coluna do pesos
* X é o vetor linha com as entradas
* b é o bias

Nesse exemplo foi utilizada a função degrau, porém outras funções de ativação podem ser utilizadas, por exemplo a sigmoid

### Treinamento

O treinamento é composto com a seguinte regra de Hebb rule: conexões que levam ao acerto no treino são fortalecidas, enquanto conexões que levam ao erro são penalizadas.

O calculo mais básico realizado para esse fortalecimento é o seguinte __weight update__:

$$w_{i+1,j+1} = w_{i,j} + \eta \cdot (y_{target,j} - y_j) \cdot x_i$$

* $w_{i+1,j+1}$ é o peso ajustado para a proxima iteração

* $w_{i,j}$ é o peso atual da iteração

* $\eta$ é a taxa de aprendizado

* $y_{target,j}$ é o valor alvo

* $y_i$ valor atual aproximado

* $x_i$ é a entrada atual

Percebe-se que o resultado do calculo é uma reta inclinada, para classificações linearmente separaveis os Perceptrons funcionam muito bem. Porém, para padrões complexos pode nao convergir.

```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.linear_model import Perceptron

iris = load_iris(as_frame=True)

X = iris.data[["petal length (cm)", "petal width (cm)"]].values
y = (iris.target == 0)

per_clf = Perceptron(random_state=42)

per_clf.fit(X,y)

X_new = [[2,0.5],[3,1]]
y_pred = per_clf.predict(X_new)
y_pred
```

### Fraquezas

Inicialmente muitos pesquisadores abandonaram os perceptrons devido ao problema de resolução de problemas simples. Um fato é que uma simples camada de perceptron tem dificuldades em realizar uma atividade de classificação ou exclusivo (XOR).

Porem essa dificuldade pode ser superada ao adicionar mais camadas:

<img src="./image5.png" width="300" height="400">

Aplicando entrada (0,0) e (1,1) produz saida 0 e aplicando (1,0) e (0,1) produz saida 1 

* sendo x1 = 0, x2 = 0
* camada 1 :

* neuronio 1 -> $[1,1]^T$*$[0,0]$ - $\frac{3}{2}$ = - $\frac{3}{2}$ 

* neuronio 2 -> $[1,1]^T$*$[0,0]$ - $\frac{1}{2}$ = - $\frac{1}{2}$ 

* camada 2 : Aplicando uma função degrau

* saida -> $[-1,1]^T$*$[-\frac{3}{2}, -\frac{1}{2}]$ - $\frac{1}{2}$ = - $\frac{2}{5}$

* porém: - $\frac{2}{5}$ < 0  = 0
